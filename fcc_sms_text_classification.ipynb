```
# get data files
!wget https://cdn.freecodecamp.org/project-data/sms/train-data.tsv
!wget https://cdn.freecodecamp.org/project-data/sms/valid-data.tsv

train_file_path = "train-data.tsv"
test_file_path = "valid-data.tsv"
```

```
import tensorflow as tf
import keras

print("TensorFlow:", tf.__version__)
print("Keras:", keras.__version__)
print("Keras location:", keras.__file__)
print("Has Sequential:", hasattr(keras, "Sequential"))
```

```
# 3

# Load train and validation data
train_data = pd.read_csv(
    train_file_path,
    sep='\t',
    header=None,
    names=['label', 'message']
)

test_data = pd.read_csv(
    test_file_path,
    sep='\t',
    header=None,
    names=['label', 'message']
)

# Convert labels:
# ham  = 0
# spam = 1
train_data['label'] = train_data['label'].map({
    'ham': 0,
    'spam': 1
})

test_data['label'] = test_data['label'].map({
    'ham': 0,
    'spam': 1
})

train_data.head()
```

```
# 4

train_messages = train_data['message'].values
train_labels = train_data['label'].values

test_messages = test_data['message'].values
test_labels = test_data['label'].values
```

```
import tensorflow as tf
import numpy as np

max_tokens = 10000
sequence_length = 100

vectorize_layer = tf.keras.layers.TextVectorization(
    max_tokens=max_tokens,
    output_mode="int",
    output_sequence_length=sequence_length
)

vectorize_layer.adapt(train_messages)

print("vectorize_layer berhasil dibuat")
```

```
# 6

model = keras.Sequential([
    vectorize_layer,

    keras.layers.Embedding(
        input_dim=max_tokens,
        output_dim=64
    ),

    keras.layers.Bidirectional(
        keras.layers.LSTM(32)
    ),

    keras.layers.Dense(32, activation='relu'),
    keras.layers.Dropout(0.5),

    keras.layers.Dense(1, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```
import pandas as pd
import numpy as np

train_df = pd.read_csv(
    "train-data.tsv",
    sep="\t",
    header=None,
    names=["label", "message"]
)

test_df = pd.read_csv(
    "valid-data.tsv",
    sep="\t",
    header=None,
    names=["label", "message"]
)

train_messages = train_df["message"].values
train_labels = (train_df["label"] == "spam").astype(int).values

test_messages = test_df["message"].values
test_labels = (test_df["label"] == "spam").astype(int).values

print("Train:", len(train_messages))
print("Test :", len(test_messages))
```

```
model = tf.keras.Sequential([
    vectorize_layer,

    tf.keras.layers.Embedding(
        input_dim=max_tokens,
        output_dim=64
    ),

    tf.keras.layers.Bidirectional(
        tf.keras.layers.LSTM(32)
    ),

    tf.keras.layers.Dense(32, activation="relu"),

    tf.keras.layers.Dropout(0.5),

    tf.keras.layers.Dense(1, activation="sigmoid")
])

model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```
history = model.fit(
    train_messages,
    train_labels,
    validation_data=(test_messages, test_labels),
    epochs=10,
    batch_size=32,
    verbose=1
)
```

```
def predict_message(pred_text):
    probability = float(
        model.predict(
            np.array([pred_text]),
            verbose=0
        )[0][0]
    )

    label = "spam" if probability >= 0.5 else "ham"

    return [probability, label]
```

```
def predict_message(pred_text):
    prediction = model.predict(
        np.array([pred_text], dtype=object),
        verbose=0
    )[0][0]

    label = "spam" if prediction >= 0.5 else "ham"

    return [float(prediction), label]


pred_text = "how are you doing today?"
prediction = predict_message(pred_text)
print(prediction)
```
