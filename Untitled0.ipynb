```python
import random
import matplotlib.pyplot as plt
import numpy as np


# =========================================================
# SCATTER PLOT GAME
# =========================================================

def scatter_plot_game():
    print("\n=== SCATTER PLOT GAME ===")

    difficulty = input("Difficulty (1=Easy, 2=Hard): ")

    limit = 5 if difficulty == "1" else 10

    x = random.randint(-limit, limit)
    y = random.randint(-limit, limit)

    plt.figure(figsize=(7, 7))
    plt.scatter(x, y, color="red", s=100)
    plt.xlim(-limit - 1, limit + 1)
    plt.ylim(-limit - 1, limit + 1)
    plt.axhline(0, color="black")
    plt.axvline(0, color="black")
    plt.grid(True)
    plt.title("Find the Point")
    plt.xlabel("X")
    plt.ylabel("Y")
    plt.show()

    answer_x = int(input("Enter X coordinate: "))
    answer_y = int(input("Enter Y coordinate: "))

    if answer_x == x and answer_y == y:
        print("Correct!")
    else:
        print(f"Wrong! The correct answer is ({x}, {y}).")


# =========================================================
# ALGEBRA GAME
# =========================================================

def algebra_game():
    print("\n=== ALGEBRA PRACTICE GAME ===")

    difficulty = input("Difficulty (1=Easy, 2=Hard): ")

    maximum = 10 if difficulty == "1" else 100

    problem_type = random.choice(["one", "two"])

    x = random.randint(-maximum, maximum)

    if problem_type == "one":
        operation = random.choice(["+", "-"])

        number = random.randint(-maximum, maximum)

        if operation == "+":
            result = x + number
            print(f"\nSolve: x + ({number}) = {result}")

        else:
            result = x - number
            print(f"\nSolve: x - ({number}) = {result}")

    else:
        multiplier = random.randint(2, 10)
        number = random.randint(-maximum, maximum)

        result = multiplier * x + number

        print(f"\nSolve: {multiplier}x + ({number}) = {result}")

    answer = int(input("x = "))

    if answer == x:
        print("Correct!")
    else:
        print(f"Wrong! The correct answer is {x}.")


# =========================================================
# PROJECTILE GAME
# =========================================================

def projectile_game():
    print("\n=== PROJECTILE GAME ===")

    # -------------------------
    # LEVEL 1
    # -------------------------

    print("\nLEVEL 1")
    print("Adjust a, b, and c to clear the wall.")

    wall_x = random.randint(3, 8)
    wall_height = random.randint(3, 8)

    print(f"Wall location : x = {wall_x}")
    print(f"Wall height   : {wall_height}")

    while True:
        a = float(input("\na = "))
        b = float(input("b = "))
        c = float(input("c = "))

        plot_projectile(a, b, c, wall_x, wall_height)

        height = a * wall_x ** 2 + b * wall_x + c

        print(f"Projectile height at wall: {height:.2f}")

        if height > wall_height:
            print("LEVEL 1 COMPLETE!")
            break

        print("The projectile hit the wall. Try again.")

    # -------------------------
    # LEVEL 2
    # -------------------------

    print("\nLEVEL 2")
    print("Enter a, b, and c to clear a harder wall.")

    wall_x = random.randint(4, 9)
    wall_height = random.randint(5, 10)

    print(f"Wall location : x = {wall_x}")
    print(f"Wall height   : {wall_height}")

    a = float(input("a = "))
    b = float(input("b = "))
    c = float(input("c = "))

    plot_projectile(a, b, c, wall_x, wall_height)

    height = a * wall_x ** 2 + b * wall_x + c

    print(f"Projectile height at wall: {height:.2f}")

    if height > wall_height:
        print("LEVEL 2 COMPLETE!")
    else:
        print("You failed to clear the wall.")


def plot_projectile(a, b, c, wall_x, wall_height):
    x = np.linspace(0, 10, 300)
    y = a * x ** 2 + b * x + c

    plt.figure(figsize=(8, 5))

    plt.plot(x, y, label="y = ax² + bx + c")

    plt.bar(
        wall_x,
        wall_height,
        width=0.3,
        color="black",
        label="Wall"
    )

    plt.xlim(0, 10)
    plt.ylim(0, max(12, wall_height + 3))

    plt.xlabel("X")
    plt.ylabel("Y")
    plt.title("Projectile Game")
    plt.grid(True)
    plt.legend()

    plt.show()


# =========================================================
# MAIN MENU
# =========================================================

def main():
    while True:

        print("\n" + "=" * 35)
        print("        MATH GAMES")
        print("=" * 35)

        print("1. Scatter Plot Game")
        print("2. Algebra Practice Game")
        print("3. Projectile Game")
        print("4. Exit")

        choice = input("\nChoose a game: ")

        if choice == "1":
            scatter_plot_game()

        elif choice == "2":
            algebra_game()

        elif choice == "3":
            projectile_game()

        elif choice == "4":
            print("Goodbye!")
            break

        else:
            print("Invalid choice.")


if __name__ == "__main__":
    main()
```
